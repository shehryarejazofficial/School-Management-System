School Management System

A school has students and teachers.

Add student, calculate marks, display result.

Check the marks of every subject.

Person → Student, Teacher.

Students can have different grading systems, such as regular or scholarship students.


A student has marks in 5 subjects. The system goes through all subjects, calculates total and percentage, and displays the grade.

```python
class Person :
  def __init__(self) :
    self.name = input("Please enter your name: ")
    self.age = int(input("Please enter your age: "))

class Teacher(Person) :
  def subject(self) :
    self.subject = input("Please enter the subject you teach: ")

class Student(Person) :
  def scholarship(self) :
    self.scholarship = input("Are you on scholarship? ")
    if self.scholarship == "yes" :
      print("You are eligible for a scholarship")
    else :
      print("You are not eligible for a scholarship")

  def grade(self) :
    self.english = int(input("Please enter your grade in English: "))
    self.math = int(input("Please enter your grade in Math: "))
    self.science = int(input("Please enter your grade in Science: "))
    self.history = int(input("Please enter your grade in History: "))
    self.geography = int(input("Please enter your grade in Geography: "))
    TotalGrade = self.english + self.math + self.science + self.history + self.geography
    Percentage = TotalGrade / 500 * 100

    if Percentage >= 90 :
      print("Your grade is A")
    elif Percentage >= 80 :
      print("Your grade is B")
    elif Percentage >= 70 :
      print("Your grade is C")
    elif Percentage >= 60 :
      print("Your grade is D")
    else :
      print("Your grade is F")


    print(f"Your total grade is {TotalGrade} out of 500")
    print(f"Your percentage is {Percentage}% out of 100%")

s1 = Student()
s1.scholarship()
s1.grade()




```
